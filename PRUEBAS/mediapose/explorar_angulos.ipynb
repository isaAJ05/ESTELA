```python
import numpy as np
import matplotlib.pyplot as plt

datos = np.load('angulos_sesion.npz')
print('Articulaciones guardadas:', list(datos.keys()))
```

```python
for nombre in datos.keys():
    valores = datos[nombre]
    print(f'{nombre}: {len(valores)} datos validos, rango {valores.min():.1f} a {valores.max():.1f} grados')
```

```python
plt.figure(figsize=(10, 5))
for nombre in datos.keys():
    plt.plot(datos[nombre], label=nombre)
plt.legend()
plt.xlabel('Fotograma (solo los que pasaron el filtro)')
plt.ylabel('Angulo (grados)')
plt.title('Angulos articulares registrados')
plt.show()
```
